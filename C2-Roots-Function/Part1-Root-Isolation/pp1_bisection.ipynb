# Phương pháp chia đôi (Bisection Method)

# Điều kiện:

* $(a,b)$ là khoảng cách ly nghiệm

* $f(x)$ liên tục trên $(a,b)$

* $f(a) \cdot f(b) < 0$

# Áp dụng


## 1. Tiên nghiệm

Đánh giá sai số: $\displaystyle |x_n - x^*| \leqslant b_n - a_n = \frac{b-a}{2^n} < \varepsilon$

### 1.1 Thuật toán
**Input:** Hàm số $f(x)$, khoảng cách ly nghiệm $(a,b)$, sai số $\varepsilon$.

**Output:** Nghiệm xấp xỉ $x^*$.

**Thuật toán:**
1. **Kiểm tra:** Nếu $f(a)f(b) > 0 \Rightarrow$ Dừng. Nếu $f(a)=0$ (hoặc $f(b)=0$) $\Rightarrow x^*=a$ (hoặc $b$), dừng.
2. **Khởi tạo:** Đặt $a_0 = a, b_0 = b$, vòng lặp $i=0$. Đặt biến $\text{sign} = 1$ nếu $f(a) > 0$, ngược lại $\text{sign} = -1$. Tính $\displaystyle n = \left\lfloor \log_2\left(\frac{b-a}{\varepsilon}\right) \right\rfloor + 1$.
3. **Lặp:** Khi $i < n$ thực hiện:
   - Tính $\displaystyle x_{i+1} = \frac{a_i+b_i}{2}$ và sai số $\displaystyle \Delta_i = \frac{b-a}{2^{i+1}}$. Nếu $f(x_{i+1}) = 0 \Rightarrow x^* = x_{i+1}$, kết thúc.
   - Cập nhật: Nếu $\text{sign} \cdot f(x_{i+1}) < 0 \Rightarrow a_{i+1} = a_i, b_{i+1} = x_{i+1}$. Ngược lại $\Rightarrow a_{i+1} = x_{i+1}, b_{i+1} = b_i$.
   - Tăng $i \leftarrow i+1$.
4. **Kết thúc:** Trả về nghiệm $x^* = x_n$.

### 1.2 Code

### 1.2.1 Có cộng thêm sai số làm tròn (chưa cho trước $\varepsilon$)
Từ công thức sai số Tiên nghiệm, suy ra Điều kiện dừng: $\displaystyle n > \log_2 \left( \frac{b-a}{\varepsilon} \right)$

```python
def bisection_iteration_v1(f,a,b,n,digit): # digit = số chữ số sau dấu phẩy
    # Input check
    if (f(a) * f(b) >= 0):
        print("You have not assumed right a and b\n")
        return

    # Chia đôi
    x = 0
    delta = 0
    sign_f_a = sign(f(a))

    results = []
    diff = b-a
    temp_2 = 2

    for i in range(n):
        # Tìm x mới
        x_new = (a+b) / 2.0
        delta = diff / temp_2
        temp_2 *= 2

        results.append({
            'i': i,
            'a_i': a,
            'b_i': b,
            'x_(i+1)': x_new,
            'f(a_i)': f(a),
            'f(b_i)': f(b),
            'f(x_(i+1))': f(x_new),
            'delta=(b-a)/2^(i+1)': delta
        })

        # Chuẩn bị cho lần lặp tiếp
        x = x_new
        if f(x_new) == 0:
            break
        elif (f(x_new) * sign_f_a < 0):
            b = x_new
        elif (f(x_new) * sign_f_a > 0):
            a = x_new

    # In kết quả
    df_results = pd.DataFrame(results)
    print(df_results.to_string(index=False))

    if digit == None:
        print(f"Nghiệm cần tìm là: {x}")
    else:
        total_delta = delta + 0.5 * 10**(-digit) # Tính toán sai số làm tròn
        print(f"Vậy x xấp xỉ x_{n} = {x}")
        print(f"Nghiệm cần tìm với {digit} chữ số sau dấu phẩy: {round(x, digit)}")
        print(f"Sai số tuyệt đối: {total_delta}")
```

## 2. Hậu nghiệm - Sai số tuyệt đối

Đánh giá sai số: $|x_n - x^*| \leqslant b_n - a_n = |x_n - x_{n-1}|$

### 2.1 Thuật toán
**Input:** Hàm số $f(x)$, khoảng cách ly nghiệm $(a,b)$, sai số $\varepsilon$.
**Output:** Nghiệm xấp xỉ $x^*$.
**Thuật toán:**
1. **Kiểm tra:** Nếu $f(a)f(b) > 0 \Rightarrow$ Dừng. Nếu $f(a)=0$ (hoặc $f(b)=0$) $\Rightarrow x^*=a$ (hoặc $b$), dừng.
2. **Khởi tạo:** Đặt $a_0 = a, b_0 = b$, mồi nghiệm $x_0 = 0$, vòng lặp $i=0$. Đặt biến $\text{sign} = 1$ nếu $f(a) > 0$, ngược lại $\text{sign} = -1$.
3. **Quá trình lặp:**
   - Tính $\displaystyle x_{i+1} = \frac{a_i+b_i}{2}$ và $\Delta x_i = |x_{i+1} - x_i|$.
   - Nếu $f(x_{i+1}) = 0$ hoặc $\Delta x_i < \varepsilon \Rightarrow x^* = x_{i+1}$, kết thúc thuật toán.
   - Cập nhật: Nếu $\text{sign} \cdot f(x_{i+1}) < 0 \Rightarrow a_{i+1} = a_i, b_{i+1} = x_{i+1}$. Ngược lại $\Rightarrow a_{i+1} = x_{i+1}, b_{i+1} = b_i$.
   - Tăng $i \leftarrow i+1$.

### 2.2 Code
### 2.2.1 Có cộng thêm sai số làm tròn (chưa cho trước $\varepsilon$)

```python
def bisection_iteration_v2 (f,a,b,n,digit):
    # Input check
    if (f(a) * f(b) >= 0):
        print("You have not assumed right a and b\n")
        return

    # Chia đôi
    x = 0
    delta_x = 0
    sign_f_a = sign(f(a))

    results = []
    for i in range(n):
        # Tìm x mới
        x_new = (a+b) / 2.0
        delta_x = abs(x_new - x)

        results.append({
            'i': i,
            'a_i': a,
            'b_i': b,
            'x_(i+1)': x_new,
            'f(a_i)': f(a),
            'f(b_i)': f(b),
            'f(x_(i+1))': f(x_new),
            'delta_x=|x_(i+1)-x_i|': delta_x
        })

        # Chuẩn bị cho lần lặp tiếp
        x =  x_new
        if f(x_new) == 0:
            break
        elif (f(x_new) * sign_f_a < 0):
            b = x_new
        elif (f(x_new) * sign_f_a > 0):
            a = x_new

    # In kết quả
    df_results = pd.DataFrame(results)
    print(df_results.to_string(index=False))

    if digit == None:
        print(f"Nghiệm cần tìm là: {x}")
    else:
        total_delta = delta_x + 0.5 * 10**(-digit) # Tính toán sai số làm tròn
        print(f"Vậy x xấp xỉ x_{n} = {x}")
        print(f"Nghiệm cần tìm với {digit} chữ số sau dấu phẩy: {round(x, digit)}")
        print(f"Sai số tuyệt đối: {total_delta}")
```

```python
f = lambda x: np.e**x - np.cos(2*x)

a = -3
b = -2

n = 30
digit = 5

bisection_iteration_v2 (f, a, b, n, digit)
```

### 2.2.2 Không cộng thêm sai số làm tròn (đã cho trước $\varepsilon$)

Từ công thức ta có: $|x_n - x^*| \leqslant |x_n - x_{n-1}| < \epsilon$ - điều kiện dừng (hậu nghiệm)

```python
def bisection_recursion_absolute (f,a,b,eps):
    # Error function
    if (f(a) * f(b) >= 0):
        print("You have not assumed right a and b\n")
        return

    # Chia đôi
    x = 0
    sign_f_a = sign(f(a))
    print(f"delta_x = {eps}")

    i = 0
    results = []
    while True:
        # Tìm x mới
        x_new = (a+b) / 2.0
        current_delta_x = abs(x_new - x)

        results.append({
            'i': i,
            'a_i': a,
            'b_i': b,
            'x_(i+1)': x_new,
            'f(a_i)': f(a),
            'f(b_i)': f(b),
            'f(x_(i+1))': f(x_new),
            'delta_x=|x_(i+1)-x_i|': current_delta_x
        })

        # Chuẩn bị cho lần lặp tiếp theo
        x = x_new
        if f(x_new) == 0:
            break
        elif (f(x_new) * sign_f_a < 0):
            b = x_new
        elif (f(x_new) * sign_f_a > 0):
            a = x_new

        # Điều kiện dừng
        if current_delta_x < eps:
            break
        else:
            i += 1

    # In kết quả
    df_results = pd.DataFrame(results)
    print(df_results.to_string(index=False))

    print(f"Giá trị cần tìm với sai số tuyệt đối {eps} là: {x}")
```

```python
f = lambda x: np.log(x) - 1 #approximate e

a = 2
b = 3

eps = 0.5 * pow(10, -7) #epsilon should be small so that the root is accurate to 7 decimal places

bisection_recursion_absolute(f, a, b, eps)
```

### 2.2.3 Tìm nghiệm thoả mãn điều kiện sai số tương đối

Từ công thức ta có: $\displaystyle \frac{|x_n - x^*|}{|x_n|} \leqslant \frac{|x_n - x_{n-1}|}{|x_n|} < \eta$ - điều kiện dừng

```python
def bisection_recursion_relative (f,a,b,eta):
    # Error function
    if (f(a) * f(b) >= 0):
        print("You have not assumed right a and b\n")
        return

    # Chia đôi
    x = 0
    sign_f_a = sign(f(a))
    print(f"sigma_x = {eta}")

    i = 0
    results = []
    while True:
        # Tìm x mới
        x_new = (a+b) / 2.0
        current_sigma_x = abs(x_new - x)/abs(x_new)

        results.append({
            'i': i,
            'a_i': a,
            'b_i': b,
            'x_(i+1)': x_new,
            'f(a_i)': f(a),
            'f(b_i)': f(b),
            'f(x_(i+1))': f(x_new),
            'sigma_x=|x_(i+1)-x_n|/|x_(i+1)': current_sigma_x
        })

        # Chuẩn bị lần lặp tiếp theo
        x = x_new
        if f(x_new) == 0:
            break
        elif (f(x_new) * sign_f_a < 0):
            b = x_new
        elif (f(x_new) * sign_f_a > 0):
            a = x_new

        # Điều kiện dừng
        if current_sigma_x < eta:
            break
        else:
            i += 1

    # Print the final result
    df_results = pd.DataFrame(results)
    print(df_results.to_string(index=False))

    print(f"Giá trị cần tìm với sai số tương đối {eta} là: {x}")
```

```python
f = lambda x: x**5 - 3*x**3 + 2*x**2 - x + 5

a = -3
b = -2

eta = 0.05 * pow(10, -2)

bisection_recursion_relative (f, a, b, eta)
```
