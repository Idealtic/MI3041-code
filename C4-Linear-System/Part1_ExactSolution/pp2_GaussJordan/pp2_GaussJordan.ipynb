### Phương pháp Gauss - Jordan
#### Input: 
- Ma trận $A_{m\times n}, B_{m\times q}$
#### Output: 
- Ma trận $X_{r\times (q+p)}$, với $p$ là số ẩn tự do, $r=n-p$ là số ẩn phụ thuộc tham số. Các phần tử trên cột q+1, q+2,...q+p đặc trưng cho sự phụ thuộc của các phần tử còn lại của $X$ vào các phần tử tự do.

#### Thuật toán

1. Lập $A|B$. Đặt tập $R = \emptyset, C = \emptyset$ (trong code tương ứng là `rowused` và `colused`) để lưu hàng, cột của phần tử khử.
2. Xét tập ứng viên $S = \{ a_{ij} \mid i \notin R, j \notin C, j < n \}$:
   - Nếu $S=\{0\}$ hoặc đã chọn đủ $\min(m, n)$ phần tử $\Rightarrow$ Chuyển **Bước 4**.
   - Chọn $a_{pq} \in S$: ưu tiên $|a_{pq}| = 1$, nếu không có chọn $|a_{pq}| = \max|S|$.
   - Thêm $p \to R$, $q \to C$.
3. Khử: $\displaystyle L_t \leftarrow L_t - \frac{a_{tq}}{a_{pq}} L_p \quad (\forall t \neq p)$. Sau đó quay lại **Bước 2**.
4. Chuẩn hoá: Với mỗi $(p, q) \in (R, C)$, thực hiện $\displaystyle L_p \leftarrow L_p / a_{pq}$.
5. Xuất nghiệm:
   - Nếu $\exists i$ mà $n \le ind[i] < n+q \Rightarrow$ Vô nghiệm, kết thúc.
   - Sắp xếp các hàng theo vị trí cột của phần tử $\neq 0$ đầu tiên tăng dần để khối ẩn cơ sở tạo thành ma trận đơn vị $I$.
   - Lập mảng `missingindex` chứa các cột của $A$ không nằm trong tập $C$. Các cột thuộc $C$ (`colused`) là **ẩn cơ sở**, các cột thuộc `missingindex` là **ẩn tự do**.
   - Dạng nghiệm: $\displaystyle X_{final} = X_0 + \sum (V_k \cdot X_k)$
     + $X_0$ ($n \times q$): Lấy $q$ cột cuối tại các hàng cơ sở, gán $0$ tại vị trí ẩn tự do.
     + $V_k$ ($n \times 1$): Xét cột ẩn tự do $k$, lấy hệ số tại các hàng cơ sở rồi **đổi dấu**. Gán $1$ tại vị trí hàng của ẩn $k$.
     + $X_k$ ($1 \times q$): Ma trận tham số tự do tuỳ ý.

# Thuật toán

**Input**: $A_{m \times n}, B_{m \times q}$.

**Output**: Tập nghiệm $X_{n \times q}$.

1. Khởi tạo $M = [A|B]$ kích thước $m \times (n+q)$. Đặt tập $\mathcal{P} = \emptyset$ để lưu vị trí phần tử khử, $\mathcal{R}$ và $\mathcal{C}$ lần lượt là tập các chỉ số hàng và cột đã chọn trong $\mathcal{P}$.

2. Đặt tập ứng viên $S = \{ m_{ij} \mid i \notin \mathcal{R}, j \notin \mathcal{C}, 1 \le j \le n \}$.

3. Nếu $S = \{0\}$ hoặc $|\mathcal{P}| = \min(m, n) \Rightarrow$ Chuyển sang Bước 6.

4. Chọn $m_{p,q} \in S$ thỏa mãn $|m_{p,q}| = 1$. Nếu không tồn tại, chọn $m_{p,q} \in S$ thỏa mãn $|m_{p,q}| = \max \{ |x| \mid x \in S \}$.

5. Thực hiện lần lượt các thao tác:

    - $\mathcal{P} = \mathcal{P} \cup \{(p, q)\}$

    - $M_{t, :} = M_{t, :} - \dfrac{m_{t,q}}{m_{p,q}} M_{p, :} \quad (\forall t \neq p)$

    - Quay lại Bước 2.

6. Lặp mỗi $(p, q) \in \mathcal{P}$, thực hiện gán: $M_{p, :} = \dfrac{M_{p, :}}{m_{p,q}}$

7. Đặt $\text{ind}_i$ là chỉ số cột $j$ đầu tiên thỏa mãn $m_{i,j} \neq 0$ trên hàng $i$. Nếu $\exists i \in \{1 \dots m\}$ sao cho $n < \text{ind}_i \le n+q \Rightarrow$ Báo vô nghiệm, dừng chương trình.

8. Sắp xếp các hàng của $M$ sao cho $\text{ind}_i$ tăng dần. Gọi $r = |\mathcal{P}|$, khi đó $r$ hàng đầu tiên chứa các ẩn cơ sở. Đặt $\mathcal{B} = \mathcal{C}$ là tập chỉ số ẩn cơ sở, $\mathcal{F} = \{1 \dots n\} \setminus \mathcal{B}$ là tập ẩn tự do.

9. Khởi tạo $X_0 \in \mathbb{R}^{n \times q}$. Thực hiện gán:

    - $X_0[\mathcal{B}, :] = M[1 \dots r, (n+1) \dots (n+q)]$

    - $X_0[\mathcal{F}, :] = 0$

10. Lặp mỗi $f \in \mathcal{F}$, khởi tạo vector $V_f \in \mathbb{R}^{n \times 1}$ và thực hiện gán:

    - $V_f[\mathcal{B}, 1] = -M[1 \dots r, f]$

    - $V_f[f, 1] = 1$

    - $V_f[\mathcal{F} \setminus \{f\}, 1] = 0$

11. Kết luận tập nghiệm $X = X_0 + \sum_{f \in \mathcal{F}} V_f X_f$ (với $X_f \in \mathbb{R}^{1 \times q}$ là tham số tự do tùy ý).


```python
from sympy import Matrix, zeros, N
import numpy as np
from IPython.display import display

#Hàm này được thêm vào để loại bỏ các giá trị nhỏ hơn 10⁻¹⁰, đảm bảo chúng được coi là 0.
def chop_small_values(mat):
    if isinstance(mat, Matrix):
        return mat.applyfunc(lambda x: 0 if abs(x) < 1e-10 else x)
    else:
        return 0 if abs(mat) < 1e-10 else mat

#Hàm tìm kiếm phần tử khử, trả về giá trị và vị trí của phần tử khử
def findpivot(AB,rowA,colA,rowused,colused):
    #Hàng và cột chưa dùng
    row = list(set(range(rowA)) - set(rowused))
    col = list(set(range(colA)) - set(colused))
    
    #Ưu tiên 1 và ưu tiên 2
    firts_prioritize_element = [AB[i, j] for i in row for j in col
                                if np.abs(float(AB[i, j])) == float(int(1))]
    if firts_prioritize_element:
        pivot = min(firts_prioritize_element, key=abs)
    else:
        second_prioritize_element = [AB[i, j] for i in row for j in col]
        pivot = max(second_prioritize_element, key=abs)
    positions = [[i, j] for i in row for j in col if AB[i, j] == pivot]
    return pivot, positions
#Hàm khử gauss_jordan
def gauss_jordan(A,B):

    A, B = Matrix(N(A)), Matrix(N(B))
    rowused = []
    colused = []
    rowA, colA = A.shape
    AB = chop_small_values(A.row_join(B))
    display(AB)
    standardization = []

    #Tìm pivot và khử theo công thức (1)
    while ((len(rowused)!=rowA) and (len(colused)!=colA)):
        pivot = findpivot(AB,rowA,colA,rowused,colused)[0]
        if (abs(pivot) < 1e-10): break
        pivotrow, pivotcol = findpivot(AB,rowA,colA,rowused,colused)[1][0]
        print(f"-> Phần tử khử (pivot) = {pivot} tại hàng {pivotrow+1}, cột {pivotcol+1}")
        for i in [x for x in range(rowA) if x!=pivotrow]:
            AB[i,:] = AB[i,:] - AB[pivotrow,:]*AB[i,pivotcol]/pivot
        rowused.append(pivotrow)
        colused.append(pivotcol)
        standardization.append([pivotrow,pivotcol])
        print("Khử")
        AB = chop_small_values(AB)
        display(AB)

    #Chuẩn hoá, chia các hàng cho pivot
    print("Chuẩn hoá")
    for i in range(len(standardization)):
        pivotrow, pivotcol = standardization[i]
        AB[pivotrow,:] = AB[pivotrow,:]/AB[pivotrow,pivotcol]
    AB = chop_small_values(AB)

    return AB, colused
#Danh sách các vị trí khác 0 đầu tiên của từng hàng, dùng để sắp xếp AB theo cấu trúc bậc thang
def indexpivot(AB):
    rowAB, colAB = AB.shape
    index = []
    for row in range(rowAB):
        for col in range(colAB):
            if abs(AB[row, col]) > 1e-10: 
                index.append(col)
                break
            if col == colAB - 1:
                index.append(col + 1)
    return index
#Giải AX=B
def gauss_jordan_solve(A, B):
    #Ma trận đã khử AB
    A, B = Matrix(N(A)), Matrix(N(B))
    AB,colused = gauss_jordan(A, B)
    print("\n=> MA TRẬN SAU KHI KHỬ VÀ CHUẨN HOÁ:")
    display(AB.applyfunc(lambda x: round(x, 4)))
    n_vars = A.shape[1]
    
    # Kiểm tra trường hợp vô nghiệm
    for i in range(AB.shape[0]):
        leading_col = -1
        for j in range(AB.shape[1]):
            if abs(AB[i, j]) > 1e-10:
                leading_col = j
                break
        if leading_col >= n_vars and leading_col < AB.shape[1]:
            print("=> KẾT LUẬN: Hệ phương trình VÔ NGHIỆM.")
            return
            
    # BỔ SUNG 3 (Đưa lên): Sắp xếp lại cột để trình bày ma trận
    free_vars = sorted(list(set(range(n_vars)) - set(colused)))
    base_vars = sorted(colused)
    n_cols_B = B.shape[1]
    b_cols_indices = list(range(n_vars, n_vars + n_cols_B))
    
    # BỔ SUNG 1: Sắp xếp hàng theo vị trí của phần tử khử (pivot) để tạo ma trận đơn vị I chuẩn
    def get_pivot_col(row_index):
        for j in base_vars:
            if abs(AB[row_index, j] - 1.0) < 1e-10: return j
        return AB.shape[1] 

    row_order = sorted(range(AB.shape[0]), key=get_pivot_col)
    AB = Matrix([AB.row(i) for i in row_order])

    if free_vars:
        idx_A_display = [x + 1 for x in base_vars]
        miss_A_display = [x + 1 for x in free_vars]
        print(f"Tập C / colused (vị trí ẩn cơ sở): {idx_A_display}")
        print(f"Missing index (vị trí bị khuyết/ẩn tự do): {miss_A_display}\n")
        
    print("\n=> MA TRẬN SAU KHI SẮP XẾP HÀNG VÀ CỘT (dạng [Ẩn cơ sở | B | Ẩn tự do]):")
    new_col_order = base_vars + b_cols_indices + free_vars
    AB_reordered = AB[:, new_col_order]
    
    # Đổi dấu các cột ẩn tự do để thể hiện thao tác chuyển vế sang phải
    for j in range(len(base_vars) + len(b_cols_indices), AB_reordered.shape[1]):
        AB_reordered[:, j] = -AB_reordered[:, j]
        
    display(AB_reordered.applyfunc(lambda x: round(x, 4)))
    
# BỔ SUNG 2: In ra nghiệm tổng quát dạng ma trận
    print("\n=> KẾT LUẬN NGHIỆM TỔNG QUÁT DƯỚI DẠNG MA TRẬN:")
    if not free_vars:
        print("Hệ có nghiệm duy nhất X =")
        X = zeros(n_vars, B.shape[1])
        for i in range(AB.shape[0]):
            leading_col = -1
            for j in range(AB.shape[1]):
                if abs(AB[i, j]) > 1e-10:
                    leading_col = j
                    break
            if leading_col != -1 and leading_col < n_vars:
                for c in range(B.shape[1]):
                    X[leading_col, c] = AB[i, n_vars + c]
        display(X.applyfunc(lambda x: round(x, 4)))
    else:
        n_cols_B = B.shape[1] 
        
        # Xác định các hàng cơ sở (không phải hàng toàn số 0)
        base_rows = []
        for i in range(AB.shape[0]):
            for j in colused:
                if abs(AB[i, j] - 1.0) < 1e-10:
                    base_rows.append((i, j))
                    break
                
        # 1. Tạo ma trận hằng số X_0
        X_0 = zeros(n_vars, n_cols_B)
        for row_idx, var_idx in base_rows:
            for c in range(n_cols_B):
                X_0[var_idx, c] = round(float(AB[row_idx, n_vars + c]), 4)
                
        # 2. Tạo danh sách các vector cột hệ số V
        V_list = []
        for j, free_col in enumerate(free_vars):
            V = zeros(n_vars, 1)
            for row_idx, var_idx in base_rows:
                # Chuyển hệ số sang vế phải nên đổi dấu
                coeff = float(AB[row_idx, free_col])
                if abs(coeff) > 1e-10:
                    V[var_idx, 0] = round(-coeff, 4)
            V[free_col, 0] = 1.0 
            V_list.append((free_col, V))

        # 3. In kết quả 
        free_vars_str = ", ".join([f"X_{i+1}" for i in free_vars])
        print(f"Hệ có vô số nghiệm phụ thuộc vào {len(free_vars)} biến tự do: {free_vars_str}")
        print(f"(Trong đó mỗi biến tự do là một ma trận tuỳ ý có kích thước 1x{n_cols_B})")
        
        expr_str = "X = X_0"
        for free_col, _ in V_list:
            expr_str += f" + (V_{free_col+1} * X_{free_col+1})"
        print(f"\nCÔNG THỨC PHÂN RÃ NGHIỆM: \n{expr_str}\n\n=== CHI TIẾT CÁC MA TRẬN ===")
        
        print("1. Ma trận hằng số X_0 =")
        display(X_0.applyfunc(lambda x: round(x, 4)))
        
        print("\n2. Các vector hệ số cột V_i =")
        for free_col, V in V_list:
            print(f"V_{free_col+1} =")
            display(V.applyfunc(lambda x: round(x, 4)))
            
    return
```
