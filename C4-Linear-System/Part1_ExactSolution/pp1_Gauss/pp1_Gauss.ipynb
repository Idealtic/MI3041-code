### Phương pháp khử Gauss
#### Input: 
- Ma trận $A_{m\times n}, B_{m\times q}$
#### Output: 
- Ma trận $X_{r\times (q+p)}$, với $p$ là số ẩn tự do, $r=n-p$ là số ẩn phụ thuộc tham số. Các phần tử trên cột q+1, q+2,...q+p đặc trưng cho sự phụ thuộc của các phần tử còn lại của $X$ vào các phần tử tự do.
#### Thuật toán
**I. Quy trình thuận**
1. Lập ma trận mở rộng $A|B$. Đặt mảng $ind$ lưu vị trí cột của phần tử $\neq 0$ đầu tiên ở mỗi hàng ($ind[i] = n+q$ nếu hàng $i$ toàn $0$).
2. Sắp xếp các hàng theo $ind$ tăng dần.
3. Duyệt $i$ từ $1 \to m$, đặt $j = ind[i]$:
   - Nếu $j \ge n$ hoặc $i=m \Rightarrow$ Kết thúc quy trình thuận.
   - Nếu $j < n \Rightarrow$ Khử: $\displaystyle L_k \leftarrow L_k - \frac{a_{kj}}{a_{ij}} L_i \quad (\forall k > i)$. Cập nhật $ind$ và sắp xếp lại.

**II. Quy trình nghịch**
1. Nếu $\exists i$ mà $n \le ind[i] < n+q \Rightarrow$ Vô nghiệm, kết thúc.
2. Lập mảng `missingindex` chứa các cột của $A$ (từ $0$ đến $n-1$) không nằm trong $ind$. Các cột thuộc $ind$ (với $ind[i] < n$) là **ẩn cơ sở**, các cột thuộc `missingindex` là **ẩn tự do**.
3. Tách hệ: Đưa $p$ cột ẩn tự do sang vế phải (đổi dấu); giữ $r$ cột ẩn cơ sở ở vế trái. Ta thu được ma trận mới biểu diễn hệ đã tách có dạng `[Ẩn cơ sở | B | Ẩn tự do]`.
4. Thế ngược giải $X_{r \times (q+p)}$: Tại cột $c$, duyệt $i$ từ $r \to 1$:
   $$ \displaystyle x_{ic} = \frac{1}{a_{ii}} \left( b_{ic} - \sum_{k=i+1}^{r} a_{ik}x_{kc} \right) $$
5. Xuất nghiệm: $\displaystyle X_{final} = X_0 + \sum (V_k \cdot X_k)$
   - $X_0$ ($n \times q$): Lấy $q$ cột đầu của $X$, gán $0$ tại vị trí ẩn tự do.
   - $V_k$ ($n \times 1$): Lấy từ $p$ cột sau của $X$, gán $1$ tại vị trí hàng của ẩn $k$.
   - $X_k$ ($1 \times q$): Ma trận tham số tự do tuỳ ý.

### Phương pháp khử Gauss

**Input**: $A_{m\times n}$, $B_{m\times q}$

**Output**: $X$

**I. Quy trình thuận**

1. Khởi tạo $M = [A|B]$. Đặt $ind[i]$ là vị trí cột của phần tử $\neq 0$ đầu tiên ở hàng $i$ ($ind[i] = n+q$ nếu hàng $i$ toàn $0$).

2. Sắp xếp các hàng của $M$ theo $ind[i]$ tăng dần.

3. Lặp $i$ từ $1$ đến $m$:

    3.1. Đặt $j = ind[i]$

    3.2. Nếu $j \ge n$ hoặc $i=m \Rightarrow$ Chuyển sang Quy trình nghịch.

    3.3. Trái lại, lặp $k$ từ $i+1$ đến $m$:

    - $L_k = L_k - \dfrac{m_{kj}}{m_{ij}} L_i$

    3.4. Cập nhật $ind[i]$ và sắp xếp lại $M$ theo $ind[i]$ tăng dần.

**II. Quy trình nghịch**

1. Nếu $\exists i \in [1, m]$ thỏa mãn $n \le ind[i] < n+q \Rightarrow$ Báo vô nghiệm, dừng chương trình.

2. Đặt $\mathcal{B} = \{ind[i] \mid ind[i] < n\}$, $\mathcal{F} = \{1, 2, \dots, n\} \setminus \mathcal{B}$, $p = |\mathcal{F}|$, $r = n - p$.

3. Cập nhật $M$: Chuyển $p$ cột thuộc $\mathcal{F}$ sang vế phải (đổi dấu), giữ $r$ cột thuộc $\mathcal{B}$ ở vế trái.

4. Khởi tạo $X_{r \times (q+p)}$. Lặp cột $c$ từ $1$ đến $q+p$ và hàng $i$ từ $r$ lùi về $1$:

    - $x_{ic} = \dfrac{1}{m_{ii}} \left( m_{ic} - \displaystyle\sum\limits_{k=i+1}^{r} m_{ik}x_{kc} \right)$

5. Gán $X_0 \in \mathbb{R}^{n \times q}$ gồm $q$ cột đầu của $X$, bổ sung $0$ tại các vị trí hàng thuộc $\mathcal{F}$.

6. Lặp mỗi $f \in \mathcal{F}$, gán $V_f \in \mathbb{R}^{n \times 1}$ từ cột tương ứng trong $p$ cột sau của $X$, bổ sung $1$ tại hàng $f$.

7. Tính và in ra $X = X_0 + \displaystyle\sum\limits_{f \in \mathcal{F}} V_f X_f$ (với $X_f \in \mathbb{R}^{1 \times q}$ là tham số tự do tuỳ ý).

```python
from sympy import Matrix, zeros, N
import numpy as np
from IPython.display import display

# Hàm này được thêm vào để loại bỏ các giá trị nhỏ hơn 10⁻¹⁰, đảm bảo chúng được coi là 0.
def chop_small_values(mat):
    if isinstance(mat, Matrix):
        return mat.applyfunc(lambda x: 0 if abs(x) < 1e-10 else x)
    else:
        return 0 if abs(mat) < 1e-10 else mat
#Danh sách các vị trí khác 0 đầu tiên của từng hàng, dùng để sắp xếp AB theo cấu trúc bậc thang
def indexpivot(AB):
    rowAB, colAB = AB.shape
    index = []
    for row in range(rowAB):
        for col in range(colAB):
            if abs(AB[row, col]) > 1e-10: 
                index.append(col)
                break
            if col == colAB - 1:
                index.append(col + 1)
    return index
    
#Vị trí các cột có phần tử ở đường chéo chính bằng 0
def missingindex(A):
    colA = A.shape[1]
    index = indexpivot(A)
    return sorted(list(set(range(colA)) - set(index)))
    
#Khử gauss
def gauss(A, B):
    A, B = Matrix(N(A)), Matrix(N(B))
    AB = chop_small_values(A.row_join(B))
    rowAB, colAB = AB.shape
    
    #Sắp xếp lại AB theo cấu trúc bậc thang trong trường hợp input không phải là bậc thang    
    index = indexpivot(AB)
    sorted_order = sorted(range(len(index)), key=lambda k: index[k])
    index = sorted(index)
    AB = chop_small_values(AB[list(sorted_order), :])
    display(AB)

    #Khử Gauss (Vừa khử, vừa sắp xếp lại AB theo cấu trúc bậc thang)
    for i in range(len(index)):
        if index[i] != colAB:
            print(f"-> Phần tử khử (pivot) = {AB[i, index[i]]} tại hàng {i+1}, cột {index[i]+1}")
            for j in range(i + 1, len(index)):
                AB[j, :] = chop_small_values(Matrix(AB[j, :] - AB[i, :] / AB[i, index[i]] * AB[j, index[i]]))
            index = indexpivot(AB)
            sorted_order = sorted(range(len(index)), key=lambda k: index[k])
            index = sorted(index)
            AB = chop_small_values(AB[list(sorted_order), :])
            print("Bước khử")
            display(AB)

    # Kiểm tra trường hợp vô số nghiệm để in ra index và missingindex
    A_new = AB[:, :A.shape[1]]
    idx_A = indexpivot(A_new)
    miss_A = missingindex(A_new)
    is_no_solution = False
    for i in indexpivot(AB):
        if i >= A.shape[1] and i < AB.shape[1]:
            is_no_solution = True
            break
            
    if not is_no_solution and miss_A:
        idx_A_display = [x + 1 for x in idx_A]
        miss_A_display = [x + 1 for x in miss_A]
        print(f"Index (vị trí ẩn cơ sở): {idx_A_display}")
        print(f"Missing index (vị trí bị khuyết/ẩn tự do): {miss_A_display}\n")

    print("Ma trận sau khi khử:")
    display(AB)
    return AB

#Giải AX=B
def gauss_solve(A, B):
    # Ma trận đã khử AB
    A, B = Matrix(N(A)), Matrix(N(B))
    AB = gauss(A, B)

    # Kiểm tra trường hợp vô nghiệm
    for i in indexpivot(AB):
        if i >= A.shape[1] and i < AB.shape[1]:
            print("=> KẾT LUẬN: Hệ phương trình VÔ NGHIỆM.")
            return 

    # Ma trận A, B mới từ ma trận đã khử    
    A_new = AB[:, :A.shape[1]]
    B_new = AB[:, A.shape[1]:]

    # Chuyển những cột có phần tử trên đường chéo chính bằng 0 sang B
    missindex = missingindex(A_new)
    free_vars = missindex
    
    if free_vars:
        for i in free_vars:
            # Chuyển cột biến tự do sang vế phải (đổi dấu)
            move_col = -1 * A_new[:, i]
            B_new = B_new.row_join(move_col)
            
        # Giữ lại các cột của biến cơ sở trong ma trận A
        keep_cols = [i for i in range(A_new.shape[1]) if i not in free_vars]
        A_new = A_new[:, keep_cols]

    if free_vars:
        print("\n=> MA TRẬN SAU KHI TÁCH HỆ (dạng [Ẩn cơ sở | B | Ẩn tự do]):")
    else:
        print("\n=> MA TRẬN SAU KHI TÁCH HỆ (dạng [Ẩn cơ sở | B]):")
    display(A_new.row_join(B_new).applyfunc(lambda x: round(x, 4)))

    # Tính toán ma trận X bằng phép thế ngược
    X = zeros(A_new.shape[1], B_new.shape[1])
    rowX, colX = X.shape
    index = indexpivot(A_new)
    
    for col in range(colX):
        for row in range(rowX - 1, -1, -1):
            sum_val = sum(A_new[row, k] * X[k, col] for k in range(index[row], rowX))
            X[row, col] = chop_small_values((B_new[row, col] - sum_val) / A_new[row, index[row]])
    
    # ==========================================
    # BỔ SUNG: IN RA NGHIỆM TỔNG QUÁT DẠNG MA TRẬN
    # ==========================================

    print("\n=> KẾT LUẬN NGHIỆM TỔNG QUÁT DƯỚI DẠNG MA TRẬN:")
    if not free_vars:
        print("Hệ có nghiệm duy nhất X =")
        display(X.applyfunc(lambda x: round(x, 4)))
    else:
        n_vars = A.shape[1] # Số biến gốc (số hàng ma trận X)
        n_cols_B = B.shape[1] # Số cột ma trận B
        
        # 1. Tạo ma trận hằng số X_0
        X_0 = zeros(n_vars, n_cols_B)
        for i in range(len(keep_cols)):
            var_idx = keep_cols[i]
            for c in range(n_cols_B):
                X_0[var_idx, c] = round(float(X[i, c]), 4)
                
        # 2. Tạo danh sách các vector cột hệ số V
        V_list = []
        for j, free_col in enumerate(free_vars):
            V = zeros(n_vars, 1)
            for i in range(len(keep_cols)):
                var_idx = keep_cols[i]
                V[var_idx, 0] = round(float(X[i, n_cols_B + j]), 4)
            V[free_col, 0] = 1.0 # Giá trị 1 tại hàng của biến tự do tương ứng
            V_list.append((free_col, V))

        # 3. In kết quả theo cấu trúc toán học
        free_vars_str = ", ".join([f"X_{i+1}" for i in free_vars])
        print(f"Hệ có vô số nghiệm phụ thuộc vào {len(free_vars)} hàng tự do: {free_vars_str}")
        print(f"(Trong đó mỗi hàng tự do là một ma trận tuỳ ý có kích thước 1x{n_cols_B})")
        
        expr_str = "X = X_0"
        for free_col, _ in V_list:
            expr_str += f" + (V_{free_col+1} * X_{free_col+1})"
        print(f"\nCÔNG THỨC PHÂN RÃ NGHIỆM: \n{expr_str}\n\n=== CHI TIẾT CÁC MA TRẬN ===")
        
        print("1. Ma trận hằng số X_0 =")
        display(X_0)
        
        print("\n2. Các vector hệ số cột V_i =")
        for free_col, V in V_list:
            print(f"V_{free_col+1} =")
            display(V)
            
    return
```

```python
A = Matrix(np.loadtxt("20252_gk_A.txt"))
print("A =")
display(A)
print("B =")
B = Matrix(np.loadtxt("20252_gk_B.txt"))
display(B)
print("A|B=")
gauss_solve(A,B)
```
